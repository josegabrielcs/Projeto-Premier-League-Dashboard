```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/raw/league_standings/standings.csv", skiprows=1, encoding='utf-8')
```

```python
df.head(5)
```

```python
# Tratar string do notes tirando símbolos
df['Notes'] = df['Notes'].str.replace('→ ', '')
df.head(5)
```

```python
# Dividir os dados da coluna Top Team Scorer
df[['Top Scorer Name', 'Top Scorer Goals']] = df['Top Team Scorer'].str.split(' - ', expand=True)
df = df.drop(columns = 'Top Team Scorer')
df.head(5)
```

```python
# Tratar artilheiros empatados trazendo apenas um nome
df['Top Scorer Name'] = df['Top Scorer Name'].str.split().str[:2].str.join(' ')
df.head(5)
```

```python
# Mudar nome de coluna
df = df.rename(columns={
    'Attendance' : 'Avg_Attendance',
    'Pts/MP' : 'Pts_MP',
    'Notes' : 'Qualification',
    'W' : 'Wins',
    'D' : 'Draws',
    'L' : 'Loses'
})
df = df.rename(columns={'Pts/MP' : 'Pts_MP'})
df = df.rename(columns={'Notes' : 'Qualification'})
df.head(5)
```

```python
# Checar tipo de dados e mudar o necessário
df['Top Scorer Goals'] = pd.to_numeric(df['Top Scorer Goals'])
df.dtypes
```

```python
with pd.option_context('display.max_columns', None):
    display(df.head(30))
```

```python
df.to_csv("../Data/clean/league_standings/League_standings_clean.csv", index=False)
```
