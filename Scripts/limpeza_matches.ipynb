```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/raw/matches_stats/Arsenal_matches.csv", skiprows=1, encoding='utf-16')
```

```python
df.head(5)
```

```python
# Excluir colunas sem utilização
df = df.drop(columns= ['Match Report', 'Notes'])
df.head(5)
```

```python
# Tratar strings de horário e Matchweek
df['Time'] = df['Time'].str[:5]
df['Round'] = df['Round'].str.replace('Matchweek ', '')
df.head(5)
```

```python
# Mudar tipo de dado das datas e round
df['Date'] = pd.to_datetime(df['Date'])
df['Round'] = pd.to_numeric(df['Round'])
df.dtypes
```

```python
df = df.sort_values(by="Round")
with pd.option_context('display.max_columns', None):
    display(df.head(38))
```

```python
df.to_csv("../Data/clean/matches_stats/Arsenal_matches_clean.csv", index=False)
```
