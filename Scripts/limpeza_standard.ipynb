```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/raw/standard_stats/Arsenal_standard.csv", skiprows=1,header=[0, 1], encoding='utf-16')
```

```python
df.head(5)
```

```python
# Fazer com que o DataFrame deixe de ser MultIndex e passe a ter apenas uma parte de colunas
df.columns = df.columns.droplevel(0)
df.head(5)
```

```python
# Renomear colunas e tratar colunas que ficaram duplicadas
lista_colunas = list(df.columns)
lista_colunas[16] = 'Gls_90'
lista_colunas[17] = 'Ast_90'
lista_colunas[18] = 'G+A_90'
lista_colunas[19] = 'G-PK_90'
lista_colunas[4] = 'Matches Played'
lista_colunas[7] = '90s Played'
df.columns = lista_colunas
df.head(5)
```

```python
# Excluir colunas e linhas sem utilização
df = df.drop(columns= ['Matches', '-9999'])
df = df.dropna(subset=['Min'])
df = df.dropna(subset=['Pos'])
df.tail(5)
```

```python
# Tratar strings da coluna Nation trazendo apenas a sigla e Pos tratando posições duplas 
df['Nation'] = df['Nation'].str[-3:]
df['Pos'] = df['Pos'].str[:2]
df.head(5)
```

```python
# Mudar tipo dos dados
conversoes = {
    'Age': 'Int64',
    'Min': 'Int64',
    'Gls': 'Int64',
    'Ast': 'Int64',
    'G+A': 'Int64',  
    'G-PK': 'Int64', 
    'PK': 'Int64',
    'PKatt': 'Int64',
    'CrdY': 'Int64',
    'CrdR': 'Int64',
}
df = df.astype(conversoes)
df.dtypes
```

```python
df.to_csv("../Data/clean/standard_stats/Arsenal_standard_clean.csv", index=False)
```
